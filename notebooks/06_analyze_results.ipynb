Chấm câu trả lời theo quy tắc đã định nghĩa, gộp các thứ tự lựa chọn và ước lượng chênh lệch bằng bootstrap theo cụm ảnh. Chỉ bước này đọc nhãn chấm điểm; không dùng nhãn để sinh mô tả hoặc trả lời.

```python
import os
import sys
from pathlib import Path

source_root = Path(os.environ.get("MEDVQA_SOURCE_ROOT", ".")).resolve()
if not (source_root / "medvqa").is_dir():
    raise FileNotFoundError("Set MEDVQA_SOURCE_ROOT to the extracted source directory.")
if str(source_root) not in sys.path:
    sys.path.insert(0, str(source_root))
```

```python
RUNS_DIR = source_root / "runs"
RESULTS_DIR = source_root / "results"
```

```python
from medvqa.analysis.run_all import analyze

result = analyze(str(RUNS_DIR), str(RESULTS_DIR))
print(result["integrity"]["status"])
```

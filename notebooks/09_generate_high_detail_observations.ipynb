# Sinh mô tả ở mức chi tiết cao

Chạy trên CPU sau khi lần thử đã được đối chiếu usage và ngân sách. Dùng package `medvqa` với phụ thuộc `api`. Tập đầu vào cố định 599 ảnh; phản hồi thử tương thích được dùng lại. Mặc định chỉ kiểm đầu vào. Khi chạy tiếp, dùng cùng thư mục output; không xóa yêu cầu thiếu phản hồi.


```python
import os
import sys
from pathlib import Path

source_root = Path(os.environ.get("MEDVQA_SOURCE_ROOT", ".")).resolve()
if not (source_root / "medvqa").is_dir():
    raise FileNotFoundError("Đặt MEDVQA_SOURCE_ROOT đến thư mục chứa package medvqa.")
if str(source_root) not in sys.path:
    sys.path.insert(0, str(source_root))
```

```python
import json

from medvqa import detail_batch

INVENTORY = source_root / "resources/detail_probe/high_detail_inputs.json"
IMAGES = Path("data/prep_out/images")
OUTPUT = Path("outputs/high_detail_observations")
BALANCE_USD = None
EXECUTE = False

if BALANCE_USD is None:
    BALANCE_USD = float(input("Số dư API hiện tại (USD): "))
inventory = detail_batch.read(INVENTORY)
forecast = detail_batch.preflight(inventory, IMAGES, BALANCE_USD)
print(json.dumps(forecast, ensure_ascii=False, indent=2))
```

```python
import getpass

if EXECUTE:
    api_key = os.environ.get("OPENAI_API_KEY") or getpass.getpass("OpenAI API key: ")
    try:
        result = detail_batch.run(INVENTORY, IMAGES, OUTPUT, BALANCE_USD, api_key)
    finally:
        api_key = None
    print(json.dumps(result, ensure_ascii=False, indent=2))
```

# Trả lời với mô tả chi tiết cao — L

Chạy trên **Kaggle T4×2** với package `medvqa`. Cố định 988 câu, 542 ảnh và 1.824 lượt trả lời. Không có lời gọi GPT trong notebook này. Chọn thư mục ảnh PNG chuẩn hóa và thư mục lưu kết quả được giữ lại sau khi runtime ngắt. Đặt `INSTALL=True` nếu cần cài môi trường của lượt chạy gốc; khởi động lại runtime nếu trình cài yêu cầu, rồi chạy lại với `INSTALL=False`.


```python
import os
import sys
from pathlib import Path

source_root = Path(os.environ.get("MEDVQA_SOURCE_ROOT", ".")).resolve()
if not (source_root / "medvqa").is_dir():
    raise FileNotFoundError("Đặt MEDVQA_SOURCE_ROOT đến thư mục chứa package medvqa.")
if str(source_root) not in sys.path:
    sys.path.insert(0, str(source_root))
```

```python
import json

from medvqa.inference_environment import install_measured_environment

ANSWERER = "L"
INPUTS = source_root / "resources/detail_probe/high_detail_answer_inputs.json"
IMAGES = Path("data/prep_out/images")
OUT_ROOT = Path("outputs/runs")
INSTALL = False
EXECUTE = False

data = json.loads(INPUTS.read_text(encoding="utf-8"))
if INSTALL:
    install_measured_environment(data, ANSWERER)
```

```python
from medvqa import high_detail_answers as runner

_, _, units = runner.preflight(INPUTS, IMAGES, ANSWERER)
environment = runner.verify_environment(data, ANSWERER)
print("Lượt trả lời:", len(units))
print(environment)
```

```python
if EXECUTE:
    result = runner.run(INPUTS, IMAGES, OUT_ROOT, ANSWERER)
    print("Còn lại:", result.get("remaining_in_shard", 0))
    print("Output:", result["run_dir"])
```
